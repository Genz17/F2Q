```python
import re
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize
from matplotlib.cm import ScalarMappable

def load_series(prefix, sizes, start_idx=5):
    series = {}
    for L in sizes:
        log_file = f"./2dLOGS/{prefix}_8_4_{L}.log"
        last_numbers = []

        with open(log_file, "r", encoding="utf-8") as f:
            for line in f:
                nums = re.findall(r"-?\d+", line)
                if nums:
                    last_numbers.append(int(nums[-1]))

        y = np.asarray(last_numbers)[start_idx:]
        y = y - y[-1] + 1
        series[L] = y

    return series


sizes = range(3, 17)

series_hopping = load_series("h2D_JW", sizes, start_idx=5)
series_hubbard = load_series("h2D_BK", sizes, start_idx=5)

cmap = plt.cm.viridis
norm = Normalize(vmin=min(sizes), vmax=max(sizes))
sm = ScalarMappable(norm=norm, cmap=cmap)
sm.set_array([])

fig, axes = plt.subplots(
    1, 2,
    figsize=(12, 4.8),
    dpi=160,
    sharey=True,
    constrained_layout=True
)

datasets = [
    (series_hopping, "2D Nearest Hopping (Jordan-Wigner)"),
    (series_hubbard, "2D Nearest Hopping (Bravyi-Kitaev)")
]

for ax, (series, title) in zip(axes, datasets):
    for L in sizes:
        y = series[L]
        x = np.arange(1, len(y) + 1)   # log scale cannot start from 0
        ax.loglog(x, y, lw=2, color=cmap(norm(L)), alpha=0.95)

    ax.set_title(title, fontsize=16)
    ax.grid(True, which="both", ls=":", alpha=0.35)
    ax.tick_params(axis="both", labelsize=12)

axes[0].set_ylabel(r"$PW(H)^{(t)} - PW(H)^{(T)}$", fontsize=16)

yticks = [1, 10, 100, 1000, 10000]
yticklabels = ["0", r"$10^1$", r"$10^2$", r"$10^3$", r"$10^4$"]
for ax in axes:
    ax.set_yticks(yticks)
    ax.set_yticklabels(yticklabels)

all_y = np.concatenate(
    [np.concatenate(list(series_hopping.values())),
     np.concatenate(list(series_hubbard.values()))]
)
ymin = 1
ymax = 10 ** np.ceil(np.log10(all_y.max()))
for ax in axes:
    ax.set_ylim(ymin, ymax)

# one shared xlabel centered under both subplots
fig.supxlabel(r"Iteration number $t$", fontsize=16)

# one shared colorbar
cbar = fig.colorbar(sm, ax=axes, pad=0.02)
cbar.set_label("System size", fontsize=16)
cbar.ax.tick_params(labelsize=12)

plt.savefig("h2d.pdf", bbox_inches="tight")
plt.show()



sizes = range(3, 17)

s1 = load_series("h2D_JW", sizes, start_idx=5)
s2 = load_series("h2D_BK", sizes, start_idx=5)
s3 = load_series("hbd_JW", sizes, start_idx=5)
s4 = load_series("hbd_BK", sizes, start_idx=5)

cmap = plt.cm.viridis
norm = Normalize(vmin=2, vmax=16)
sm = ScalarMappable(norm=norm, cmap=cmap)
sm.set_array([])

fig, axes = plt.subplots(
    1, 4,
    figsize=(24, 5),
    dpi=160,
    sharey=True,
    constrained_layout=True
)

datasets = [
    (s1, "2D Nearest Hopping (Jordan-Wigner)"),
    (s2, "2D Nearest Hopping (Bravyi-Kitaev)"),
    (s3, "2D Hubbard Model (Jordan-Wigner)"),
    (s4, "2D Hubbard Model (Bravyi-Kitaev)")
]

for ax, (series, title) in zip(axes, datasets):
    for L in sizes:
        y = series[L]
        x = np.arange(1, len(y) + 1)   # log scale cannot start from 0
        ax.loglog(x, y, lw=2, color=cmap(norm(L)), alpha=0.95)

    ax.set_title(title, fontsize=21)
    #ax.grid(False, which="both", ls=":", alpha=0.35)
    ax.tick_params(axis="both", labelsize=18)

axes[0].set_ylabel(r"$PW(H)^{(t)} - PW(H)^{(T)}$", fontsize=20)

yticks = [1, 10, 100, 1000, 10000,]
yticklabels = ["0", r"$10^1$", r"$10^2$", r"$10^3$", r"$10^4$",]
for ax in axes:
    ax.set_yticks(yticks)
    ax.set_yticklabels(yticklabels)

all_y = np.concatenate(
    [np.concatenate(list(series_hopping.values())),
     np.concatenate(list(series_hubbard.values()))]
)
ymin = 1
ymax = 5 * 10 ** 4
for ax in axes:
    ax.set_ylim(ymin, ymax)

# one shared xlabel centered under both subplots
fig.supxlabel(r"Iteration number $t$", fontsize=26)

# one shared colorbar
cbar = fig.colorbar(sm, ax=axes, pad=0.02)
cbar.set_label("System size", fontsize=20)
cbar.ax.tick_params(labelsize=24)

plt.savefig("2d_convergence.pdf", bbox_inches="tight")
plt.show()
```
